```python
def income_group(income):
    if income < 300000:
        return "Low Income"
    elif income < 600000:
        return "Lower Middle"
    elif income < 1000000:
        return "Middle Income"
    elif income < 1500000:
        return "Upper Middle"
    else:
        return "High Income"

df["Income_Group"] = df["Annual_Income"].apply(income_group)
```

```python
def score_category(score):
    if score >= 800:
        return "Excellent"
    elif score >= 750:
        return "Very Good"
    elif score >= 700:
        return "Good"
    elif score >= 650:
        return "Fair"
    elif score >= 550:
        return "Poor"
    else:
        return "Very Poor"

df["Credit_Score_Category"] = df["Credit_Score"].apply(score_category)
```

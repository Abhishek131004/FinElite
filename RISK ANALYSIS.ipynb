```python
#Payment Behaviour Visualization
plt.figure(figsize=(18,12))

plt.subplot(2,3,1)
sb.histplot(df["Credit_Score"], bins=15, kde=True, color="blue")
plt.title("Credit Score Distribution")

plt.subplot(2,3,2)
sb.histplot(df["Missed_Payments"], bins=6, kde=True, color="orange")
plt.title("Missed Payments")

plt.subplot(2,3,3)
sb.histplot(df["Late_Payment_Count"], bins=9, kde=True, color="red")
plt.title("Late Payment Count")

plt.subplot(2,3,4)
sb.histplot(df["Debt_To_Income_Ratio"], bins=15, kde=True, color="purple")
plt.title("Debt To Income Ratio")

plt.subplot(2,3,5)
sb.histplot(df["Credit_Utilization"], bins=15, kde=True, color="green")
plt.title("Credit Utilization")

plt.subplot(2,3,6)
sb.countplot(x="Number_of_Defaults", data=df, color="skyblue")
plt.title("Number of Defaults")

plt.tight_layout()
plt.show()
```

```python
#Create Risk Score
df["Risk_Score"] = (
    (df["Number_of_Defaults"] > 0).astype(int) * 40 +
    (df["Missed_Payments"] >= 4).astype(int) * 20 +
    (df["Late_Payment_Count"] >= 6).astype(int) * 15 +
    (df["Credit_Score"] < 650).astype(int) * 15 +
    (df["Debt_To_Income_Ratio"] >= 0.50).astype(int) * 5 +
    (df["Credit_Utilization"] >= 80).astype(int) * 5
)
```

```python
df["Risk_Category"] = pd.cut(
    df["Risk_Score"],
    bins=[-1, 19, 39, 59, 100],
    labels=[
        "Low Risk",
        "Moderate Risk",
        "High Risk",
        "Very High Risk"
    ]
)

df["Risk_Category"].value_counts().sort_index()
```

```python
#Risk Category Analysis
df.groupby("Risk_Category")[[
    "Credit_Score",
    "Missed_Payments",
    "Late_Payment_Count",
    "Number_of_Defaults",
    "Debt_To_Income_Ratio",
    "Credit_Utilization",
    "Credit_Limit"
]].agg(["count", "mean"]).round(2)
```

```python
#High-Risk Customers
high_risk_customers = df[
    df["Risk_Category"].isin([
        "High Risk",
        "Very High Risk"
    ])
]
```

```python
high_risk_customers[[
    "Customer_ID",
    "Age",
    "Occupation",
    "Annual_Income",
    "Credit_Score",
    "Missed_Payments",
    "Late_Payment_Count",
    "Number_of_Defaults",
    "Debt_To_Income_Ratio",
    "Credit_Utilization",
    "Risk_Score",
    "Risk_Category"
]].sort_values(
    by=[
        "Risk_Score",
        "Number_of_Defaults",
        "Missed_Payments"
    ],
    ascending=False
).head(20)
```

```python
#Create Age Groups
def age_group(age):

    if age < 25:
        return "18-24"

    elif age < 35:
        return "25-34"

    elif age < 45:
        return "35-44"

    elif age < 55:
        return "45-54"

    else:
        return "55+"

df["Age_Group"] = df["Age"].apply(age_group)
```

```python
#Risk Dashboard
fig = plt.figure(figsize=(20,15))

fig.add_subplot(331)
sb.countplot(
    data=df,
    x="Risk_Category",
    color="skyblue"
)
plt.title("Risk Category")
plt.xticks(rotation=20)


fig.add_subplot(332)
sb.barplot(
    data=df.groupby(
        "Occupation",
        as_index=False
    )["Number_of_Defaults"].mean(),
    x="Occupation",
    y="Number_of_Defaults",
    color="red"
)
plt.xticks(rotation=90)
plt.title("Average Defaults by Occupation")


fig.add_subplot(333)
sb.barplot(
    data=df.groupby(
        "Occupation",
        as_index=False
    )["Missed_Payments"].mean(),
    x="Occupation",
    y="Missed_Payments",
    color="orange"
)
plt.xticks(rotation=90)
plt.title("Average Missed Payments by Occupation")


fig.add_subplot(334)
sb.barplot(
    data=df.groupby(
        "Age_Group",
        as_index=False
    )["Number_of_Defaults"].mean(),
    x="Age_Group",
    y="Number_of_Defaults",
    color="blue"
)
plt.title("Average Defaults by Age Group")


fig.add_subplot(335)
sb.barplot(
    data=df.groupby(
        "Age_Group",
        as_index=False
    )["Missed_Payments"].mean(),
    x="Age_Group",
    y="Missed_Payments",
    color="purple"
)
plt.title("Average Missed Payments by Age Group")


fig.add_subplot(336)
sb.scatterplot(
    data=df,
    x="Credit_Score",
    y="Debt_To_Income_Ratio",
    hue="Risk_Category"
)
plt.title("Credit Score vs Debt To Income")


fig.add_subplot(337)
sb.scatterplot(
    data=df,
    x="Credit_Utilization",
    y="Missed_Payments",
    hue="Risk_Category"
)
plt.title("Credit Utilization vs Missed Payments")


fig.add_subplot(338)
sb.boxplot(
    data=df,
    x="Risk_Category",
    y="Credit_Score",
    color="green"
)
plt.title("Credit Score by Risk Category")
plt.xticks(rotation=20)


fig.add_subplot(339)
sb.countplot(
    data=df,
    x="Fraud_Flag",
    hue="Risk_Category"
)
plt.title("Fraud Flag vs Risk Category")

plt.tight_layout()
plt.show()
```

```python
#Compare Risk Across Occupations
occupation_risk = df.groupby("Occupation").agg(

    Customers=("Customer_ID", "count"),

    Avg_Credit_Score=(
        "Credit_Score",
        "mean"
    ),

    Avg_Missed_Payments=(
        "Missed_Payments",
        "mean"
    ),

    Avg_Late_Payments=(
        "Late_Payment_Count",
        "mean"
    ),

    Default_Rate=(
        "Number_of_Defaults",
        lambda x: (x > 0).mean() * 100
    ),

    Avg_DTI=(
        "Debt_To_Income_Ratio",
        "mean"
    ),

    Avg_Credit_Utilization=(
        "Credit_Utilization",
        "mean"
    ),

    High_Risk_Customers=(
        "Risk_Category",
        lambda x: x.isin([
            "High Risk",
            "Very High Risk"
        ]).sum()
    )
).sort_values(
    "Default_Rate",
    ascending=False
)

occupation_risk.round(2)
```

```python
plt.figure(figsize=(12,6))

sb.barplot(
    data=occupation_risk.reset_index(),
    x="Occupation",
    y="Default_Rate",
    color="red"
)

plt.xticks(rotation=45)

plt.title("Default Rate Across Occupations")

plt.ylabel("Default Rate (%)")

plt.show()
```

```python
#Compare Risk Across Age Groups
age_risk = df.groupby("Age_Group").agg(

    Customers=("Customer_ID", "count"),

    Avg_Credit_Score=(
        "Credit_Score",
        "mean"
    ),

    Avg_Missed_Payments=(
        "Missed_Payments",
        "mean"
    ),

    Avg_Late_Payments=(
        "Late_Payment_Count",
        "mean"
    ),

    Default_Rate=(
        "Number_of_Defaults",
        lambda x: (x > 0).mean() * 100
    ),

    Avg_DTI=(
        "Debt_To_Income_Ratio",
        "mean"
    ),

    Avg_Credit_Utilization=(
        "Credit_Utilization",
        "mean"
    ),

    High_Risk_Customers=(
        "Risk_Category",
        lambda x: x.isin([
            "High Risk",
            "Very High Risk"
        ]).sum()
    )

).round(2)

age_risk
```

```python
plt.figure(figsize=(10,6))

sb.barplot(
    data=age_risk.reset_index(),
    x="Age_Group",
    y="Default_Rate",
    color="blue"
)

plt.title("Default Rate Across Age Groups")

plt.ylabel("Default Rate (%)")

plt.show()
```

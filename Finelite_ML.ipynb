# **FINELITE** (AI-powered credit card limit prediction ) **Machine Learning**
* This code builds a Regression Machine Learning project to predict a customer's Credit Limit using customer-related financial and demographic information.

**Import Required Libraries**

```python
import pandas as pd
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
```

**Load the Excel Dataset**

```python
df = pd.read_excel("../Datasets/Credir_Card_Bank.xlsx")
```

```python
pd.set_option("display.max_columns",None)
df
```

**Check Missing Values**

```python
df.isnull().sum()
```

**Check Data Types**

```python
df.dtypes
```

**Remove Unnecessary Columns**

```python
df = df.drop(columns=["Customer_ID","Monthly_Income","PAN_Verified", "KYC_Status"])
```

**Convert Categorical Columns into Numerical Columns**

```python
df=pd.get_dummies(df,columns=['Gender','Employment_Type','Occupation','Residential_Status','Fraud_Flag'],drop_first=True)
```

**Correlation Analysis**

```python
df.corr(numeric_only=True)
```

**Separate Features and Target Variable**

```python
x=df.drop(['Credit_Limit'],axis=1).values
y=df['Credit_Limit'].values
```

**Split Data into Training and Testing Sets**

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest= train_test_split(x,y,test_size=0.2)
```

# **Linear Regression**

```python
from sklearn.linear_model import LinearRegression
modellinear= LinearRegression()
modellinear.fit(xtrain,ytrain)
yplinear= modellinear.predict(xtest)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
r2linear = r2_score(ytest,yplinear)
r2linear * 100
```

```python
MAE = mean_absolute_error(ytest, yplinear)
print(MAE)
MSE= mean_squared_error(ytest, yplinear)
print(MSE)
RMSE=np.sqrt(mean_squared_error(ytest, yplinear))
print(RMSE)
```

**15-Fold Cross-Validation for Linear Regression**

```python
from sklearn.model_selection import cross_val_score
modelcvlinear = LinearRegression()
sc = cross_val_score(modelcvlinear,x,y,cv=15,scoring='r2')
print(sc)
sc.mean()
```

# **Decision Tree Regression**

```python
from sklearn.tree import DecisionTreeRegressor
modeldr = DecisionTreeRegressor()
modeldr.fit(xtrain,ytrain)
ypdr=modeldr.predict(xtest)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
r2dr = r2_score(ytest,ypdr)
print("R2 score of decision tree is",r2dr * 100)
MAE1 = mean_absolute_error(ytest, ypdr)
print("Mean absolute error for decision tree",MAE1)
MSE1= mean_squared_error(ytest, ypdr)
print("Mean squared error for decision tree",MSE1)
RMSE1=np.sqrt(mean_squared_error(ytest, ypdr))
print("Root mean squared error for decision tree",RMSE1)
```

**15-Fold Cross-Validation for Decision Tree**

```python
from sklearn.model_selection import cross_val_score
modelcvdr = DecisionTreeRegressor()
sc1 = cross_val_score(modelcvdr,x,y,cv=15,scoring='r2')
print(sc1)
sc1.mean()
```

**Hyperparameter Tuning — Decision Tree**

```python
from sklearn.model_selection import GridSearchCV 
from sklearn.tree import DecisionTreeRegressor 
tunned_parameters = { 'criterion': ['squared_error', 'friedman_mse', 'absolute_error'], 
                      'max_depth': [3, 5, 10, 15, 20, None], 
                      'min_samples_split': [2, 5, 10], 
                      'min_samples_leaf': [1, 2, 4] }
model_gscv = DecisionTreeRegressor(random_state=42)
new_model = GridSearchCV( estimator=model_gscv, param_grid=tunned_parameters, scoring='r2', cv=5 )
new_model.fit(xtrain, ytrain) 
print("Best Parameters:",new_model.best_params_) 
print("\nBest R2 Score:",new_model.best_score_) 
```

# **Random Forest Regression**

```python
from sklearn.ensemble import RandomForestRegressor 
modelrf = RandomForestRegressor( random_state=42 ) 
modelrf.fit(xtrain, ytrain) 
yprf = modelrf.predict(xtest)
```

```python
r2rf = r2_score(ytest, yprf) 
print( "R2 score of random forest is", r2rf * 100 ) 
MAE2 = mean_absolute_error( ytest, yprf )
print( "Mean absolute error for random forest", MAE2 )
MSE2 = mean_squared_error( ytest, yprf ) 
print( "Mean squared error for random forest", MSE2)
RMSE2 = np.sqrt(MSE2) 
print( "Root mean squared error for random forest", RMSE2)
```

**Cross-Validation — Random Forest**

```python
from sklearn.model_selection import cross_val_score
modelcvrf = RandomForestRegressor()
sc2 = cross_val_score(modelcvrf,x,y,cv=15,scoring='r2')
print(sc2)
sc2.mean()
```

**Random Forest Hyperparameter Tuning**

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
tunned_parameters_rf = {'n_estimators': [100, 200],
                        'max_depth': [10, None],
                        'min_samples_split': [2, 5],
                        'min_samples_leaf': [1, 2]}
model_gscv_rf = RandomForestRegressor(random_state=42,n_jobs=-1)
new_model_rf = GridSearchCV(estimator=model_gscv_rf,param_grid=tunned_parameters_rf,scoring='r2',cv=5,n_jobs=-1)
new_model_rf.fit(xtrain, ytrain)
print("Best Parameters:", new_model_rf.best_params_)
print("\nBest R2 Score:", new_model_rf.best_score_)
```

# **XGBoost Regression**

```python
from xgboost import XGBRegressor
modelxgbr = XGBRegressor( random_state=42 ) 
modelxgbr.fit(xtrain, ytrain) 
ypxgbr = modelxgbr.predict(xtest)
```

```python
r2xgbr = r2_score(ytest, ypxgbr) 
print( "R2 score of XGBoost is", r2xgbr * 100 ) 
MAE3 = mean_absolute_error( ytest, ypxgbr)
print( "Mean absolute error for XGBoost", MAE3 )
MSE3 = mean_squared_error( ytest, ypxgbr) 
print( "Mean squared error for XGBoost", MSE3)
RMSE3 = np.sqrt(MSE3) 
print( "Root mean squared error for XGBoost", RMSE3)
```

**15-Fold Cross-Validation — XGBoost**

```python
from sklearn.model_selection import cross_val_score
modelcvxgbr = XGBRegressor()
sc3 = cross_val_score(modelcvxgbr,x,y,cv=15,scoring='r2')
print(sc3)
sc3.mean()
```

**XGBoost Hyperparameter Tuning**

```python
from sklearn.model_selection import GridSearchCV
from xgboost import XGBRegressor
tunned_parameters_xgb = {'n_estimators': [100, 200],
                         'learning_rate': [0.05, 0.1],
                         'max_depth': [3, 5],
                         'subsample': [0.8, 1.0],
                         'colsample_bytree': [0.8, 1.0]}
model_gscv_xgb = XGBRegressor(objective='reg:squarederror',random_state=42,n_jobs=-1)
new_model_xgb = GridSearchCV(estimator=model_gscv_xgb,param_grid=tunned_parameters_xgb,scoring='r2',cv=5,n_jobs=-1)
new_model_xgb.fit(xtrain, ytrain)
print("Best Parameters:", new_model_xgb.best_params_)
print("\nBest R2 Score:", new_model_xgb.best_score_)
```
